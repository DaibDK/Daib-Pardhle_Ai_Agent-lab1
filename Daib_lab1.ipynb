```python
# ==============================================================================
# AI ASSIGNMENT: SIMPLE REFLEX TRAVEL AGENT (PAKISTAN TOURISM EDITION)
# ==============================================================================
# AI Course Concept:
# This program demonstrates a "Simple Reflex Agent" (Russell & Norvig).
# It uses Condition-Action Rules (if-elif-else) to map user percepts directly
# to tourism destinations and packages across Pakistan.
#
# Rule Model:
# IF (Category == X and Budget == PKR Tier) THEN (Recommend Pakistan Destination)
# ==============================================================================

def pakistan_travel_agent():
    print("=" * 65)
    print("      PAKISTAN TOURISM AI TRAVEL AGENT (RULE-BASED AI)      ")
    print("=" * 65)
    print("Available Categories: Mountain | Beach | Historical | Adventure\n")

    # 1. PERCEPTION (Sensor inputs from user)
    category = input("Select trip category (mountain/beach/historical/adventure): ").strip().lower()

    try:
        budget = float(input("Enter your total budget in PKR (Rs.) [e.g. 50000]: ").strip() or 50000)
        days = int(input("Enter trip duration in days [e.g. 4]: ").strip() or 4)
    except ValueError:
        print("\n[!] Invalid input. Using defaults: Budget = Rs. 50,000 | Duration = 4 Days")
        budget = 50000.0
        days = 4

    print("\n" + "-" * 65)
    print("AI IS EVALUATING RULES OVER PAKISTAN TOURISM KNOWLEDGE BASE...")
    print("-" * 65)

    # 2. REASONING ENGINE (Condition-Action Rules via IF-ELIF-ELSE)
    # -------------------------------------------------------------
    # RULE SET 1: MOUNTAINS & NORTHERN VALLEYS
    # -------------------------------------------------------------
    if category == "mountain":
        if budget < 30000:
            destination = "Swat Valley & Malam Jabba (KPK)"
            hotel = "Standard Riverfront Guesthouse"
            daily_cost = 4500
            activities = "Malam Jabba chairlift, Fizagat park, and fresh trout fish"
        elif budget <= 80000:
            destination = "Hunza Valley & Attabad Lake (Gilgit-Baltistan)"
            hotel = "Scenic Valley View Hotel"
            daily_cost = 11000
            activities = "Boating at Attabad Lake, visiting Baltit Fort, and Passu Cones"
        else:
            destination = "Skardu & Deosai Plains (Gilgit-Baltistan)"
            hotel = "Shangrila Resort / Serena Shigar Fort"
            daily_cost = 25000
            activities = "Upper Kachura Lake, Cold Desert safari, and Deosai plateau tour"

    # -------------------------------------------------------------
    # RULE SET 2: BEACH & COASTAL ESCAPES
    # -------------------------------------------------------------
    elif category == "beach":
        if budget < 30000:
            destination = "Hawkesbay & Clifton Beach, Karachi (Sindh)"
            hotel = "Karachi Seaside Beach Hut"
            daily_cost = 4000
            activities = "Camel riding on beach, sunset watching, and Do Darya dining"
        elif budget <= 80000:
            destination = "Kund Malir & Ormara Beach (Balochistan)"
            hotel = "Makran Coastal Beach Camp & Pods"
            daily_cost = 9500
            activities = "Drive on Makran Coastal Highway, Princess of Hope, and stargazing"
        else:
            destination = "Gwadar Hammerhead & Marine Drive (Balochistan)"
            hotel = "Gwadar Luxury Cliff-side Resort"
            daily_cost = 22000
            activities = "Private yacht ride, deep-sea fishing, and Hammerhead cliff view"

    # -------------------------------------------------------------
    # RULE SET 3: HISTORICAL & CULTURAL HERITAGE
    # -------------------------------------------------------------
    elif category == "historical":
        if budget < 30000:
            destination = "Taxila Ancient Buddhist Ruins & Museum (Punjab)"
            hotel = "Heritage Rest House Taxila"
            daily_cost = 3500
            activities = "Dharmarajika Stupa, Taxila Museum, and famous local Chapli Kabab"
        elif budget <= 80000:
            destination = "Walled City of Lahore (Punjab)"
            hotel = "Old City Heritage Boutique Hotel"
            daily_cost = 8500
            activities = "Badshahi Mosque, Shahi Qila, Shahi Hammam, and Haveli Food Street"
        else:
            destination = "Multan & Bahawalpur Palaces (Punjab)"
            hotel = "Grand Luxury Heritage Hotel"
            daily_cost = 18000
            activities = "Noor Mahal palace tour, Derawar Fort in Cholistan, and Sufi shrines"

    # -------------------------------------------------------------
    # RULE SET 4: ADVENTURE & SPORTS
    # -------------------------------------------------------------
    elif category == "adventure":
        if budget < 30000:
            destination = "Khanpur Dam & Margalla Hills (Islamabad/KPK)"
            hotel = "Lakeview Adventure Camp"
            daily_cost = 4000
            activities = "Jet skiing, cliff diving, parasailing, and Trail 3 hiking"
        elif budget <= 80000:
            destination = "Fairy Meadows & Nanga Parbat Base (Gilgit-Baltistan)"
            hotel = "Alpine Wooden Cabin at Fairy Meadows"
            daily_cost = 12000
            activities = "Thrilling 4x4 Jeep ride, forest trekking, and Nanga Parbat reflection view"
        else:
            destination = "Kumrat Valley & Katora Lake Trek (KPK)"
            hotel = "Kumrat Luxury Glamping Pods"
            daily_cost = 20000
            activities = "Dense deodar forest exploration, waterfall safari, and alpine lake trek"

    # -------------------------------------------------------------
    # DEFAULT RULE (Fallback if category is unrecognized)
    # -------------------------------------------------------------
    else:
        print("[!] Unrecognized category. Defaulting to Federal Capital Tour.")
        destination = "Islamabad & Murree Hills (Federal/Punjab)"
        hotel = "Islamabad City Hotel"
        daily_cost = 6000
        activities = "Faisal Mosque, Daman-e-Koh, Monal dinner, and Mall Road stroll"

    # 3. ACTION (Outputs recommendation and financial assessment)
    total_cost = daily_cost * days

    print("=" * 65)
    print("                PAKISTAN AI TRAVEL RECOMMENDATION               ")
    print("=" * 65)
    print(f"[>] Recommended Destination:  {destination}")
    print(f"[>] Recommended Stay:         {hotel}")
    print(f"[>] Key Activities:           {activities}")
    print(f"[$] Estimated Daily Cost:     Rs. {daily_cost:,} per day")
    print(f"[$] Total Cost ({days} Days):      Rs. {total_cost:,}")
    print("-" * 65)

    # Budget satisfaction check
    if total_cost <= budget:
        savings = budget - total_cost
        print(f"[+] Budget Status:  APPROVED! Within your budget.")
        print(f"[+] Extra Savings:  Rs. {savings:,.2f} remaining for shopping & souvenirs.")
    else:
        deficit = total_cost - budget
        print(f"[-] Budget Status:  OVER BUDGET by Rs. {deficit:,.2f}!")
        print("[*] AI Advice:      Reduce your trip duration or select a budget destination.")
    print("=" * 65 + "\n")


if __name__ == "__main__":
    pakistan_travel_agent()
```
